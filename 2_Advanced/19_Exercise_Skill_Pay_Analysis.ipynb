```python
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
from adjustText import adjust_text

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x:ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DA_IN = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'India')].copy()
df_DA_IN = df_DA_IN.dropna(subset=['salary_year_avg'])

df_DA_IN
```

```python
df_DA_IN = df_DA_IN.explode('job_skills')

df_DA_IN['job_skills']
```

```python
df_DA_IN_group = df_DA_IN.groupby('job_skills')['salary_year_avg'].agg(
    ['count','median']
)

df_DA_top_pay = df_DA_IN_group.sort_values(by='median',ascending=False).head(10)
df_DA_skills = df_DA_IN_group.sort_values(by='count',ascending=False).head(10).sort_values(by='median',ascending=False)
df_DA_top_pay,df_DA_skills
```

```python
fig,ax=plt.subplots(2,1)

df_DA_top_pay.plot(kind='barh',y='median',ax=ax[0],title='Top 10 Highest-Paying Skills for Data Analysts in India',legend=False)
ax[0].set_ylabel('')
ax[0].invert_yaxis()

df_DA_skills.plot(kind='barh',y='median',ax=ax[1],title='Top 10 Most Demanded Skills for Data Analysts in India',legend=False)
ax[1].invert_yaxis()
ax[1].set_ylabel('')

ax[0].set_xlim(0,175000)
ax[1].set_xlim(0,175000)

plt.tight_layout()
```
